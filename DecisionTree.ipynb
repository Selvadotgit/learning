## Decision Tree Implementation

```python
class Node:
    def __init__(self,feature=None,threshold =None,left=None,right=None,const_value=None):
        self.feature=feature
        self.threshold = threshold
        self.left=left
        self.right=right
        self.const_value=const_value
```

## Gini Index Calculation

```python
def _gini(self,y):
    _,count = np.unique(y,return_counts=True)
    prop = count/len(y)
    return 1-np.sum(prop**2)
```

## split the data

```python
def best_split(self,X,y):
    n_samples,features = X.shape
    max_gain = -np.inf
    best_feature=None
    best_threshold=None
    for feature in features:
        value = X[:, feature]
        parent_gini = self._gini(y)
        thresholds = np.sum(value[:-1]+ value[1:])/2.0
        for threshold in thresholds:
            left_mask = (X[:,feature]<threshold)
            right_mask = ~left_mask
            y_left = y[left_mask]
            y_right = y[right_mask]
            gini_left = self._gini(y_left)
            gini_right =self._gini(y_right)
            n_left = len(y_left)
            n_right=len(y_right)
            if n_left ==0 or n_right==0:
                continue
            weighted_gini = (n_left*gini_left/n_samples) + (n_right*gini_right/n_samples)
            information_gain  = parent_gini - weighted_gini
            if  max_gain<information_gain:
                information_gain =max_gain
                best_feature =feature
                best_threshold = threshold
    return (information_gain,best_feature,best_threshold)
    
            
            
            
            
            
            
            
    
    
```
