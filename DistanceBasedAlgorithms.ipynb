```python
import numpy as np
```

## Equalidean distance

```python
v1 = np.random.randint(1,10,size=4)
v2 = np.random.randint(1,10,size=4)
```

```python
l2_norm = np.sqrt(np.sum((v1-v2)**2))
```

```python
l2_norm
```

## Manhattan Distance

```python
l1_norm= np.sum(np.abs(v1-v2))
```

```python
l1_norm
```

## Minkowski Distance
Generalization of distance formula

```python
p = 2
m_d = np.power(np.sum((np.abs(v1-v2)**p)),1/p)
```

```python
m_d
```

## Cosine Similarity

```python
cos_theta = (v1@v2)/(np.linalg.norm(v1)*np.linalg.norm(v2))
```

```python
cos_theta
```

# KNN Alogrithm

```python
class KNN:
    def __init__(self,K):
        self.K=K
        self.X =None
        self.y =None
    def equ_distance(self,x_new):
        return np.linalg.norm(self.X - x_new, axis=1)    
    def fit(self,X,y):
        self.X =X
        self.y=y
    def predict(self,X):
        dist = self.equ_distance(X)
        sorted_y = self.y[dist.argsort()]
        return sorted_y[:self.K]
class KNNClassifier(KNN):
    def __init__(self,K):
        super().__init__(K)
    def predict(self,X):
        pred = []
        for x in X:
            sorted_y = super().predict(x)
            pred.append(np.bincount(sorted_y).argmax())
        return np.array(pred)
class KNNRegressor(KNN):
    def __init__(self,K):
        super().__init__(K)
    def predict(self,X):
        pred = []
        for x in X:
            sorted_y = super().predict(x)
            pred.append(np.mean(sorted_y))
        return np.array(pred)
        

        
        
        
```
